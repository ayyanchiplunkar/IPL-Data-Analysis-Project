# IPL Data Analysis Project
This project analyzes IPL match data using Python, Pandas, NumPy, and Matplotlib to discover insights about teams, batsmen, bowlers, and match results.

```python
# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

##  Load Dataset

```python
matches = pd.read_csv('matches.csv')
deliveries=pd.read_csv("deliveries.csv")
```

##  Data Exploratio

```python
matches.head()
```

```python
matches.shape
```

```python
matches.columns
```

## Most Successful IPL Teams

```python
teams=matches ['winner'].value_counts().plot(kind='bar')

plt.title ( 'IPL Team wins')
plt.xlabel('Teams')
plt.ylabel('wins')

plt.show()
```

## MOST TOSS WINNING TEAMS

```python
toss_win=matches['toss_winner'].value_counts().plot(kind='bar')

plt.title('TOSS WINNER')
plt.xlabel('Teams')
plt.ylabel('Wins')

plt.show()
           
```

## MOSR MAN OF THE MATCH AWARDS

```python
top_player_of_match=matches['player_of_match'].value_counts().head(10)
top_player_of_match.plot(kind='bar')

plt.title('Top player of match winner')
plt.xlabel('Players')
plt.ylabel('Awards')
plt.yticks(range(0,25))
plt.show()
```

## RESLUTS

```python
results=matches['result'].value_counts().plot(kind='bar',color='green')
plt.title('Match Results')
plt.xlabel('Result')
plt.grid()
plt.show()
```

## TOSS WINNER VS MATCH WINNER

```python
toss_match=(matches['toss_winner']==matches['winner']).value_counts()
toss_match.index=['won','lost']
toss_match.plot(kind='bar',color='red')

plt.title('Toss winner vs match winner')

plt.xlabel('Result')
plt.ylabel('count')
plt.grid()
plt.show()
```

## TOTAL RUNS BY IPL TEAMS

```python
team_run=deliveries.groupby('batting_team')['total_runs'].sum()
team_run.plot(kind='barh',color ='purple')
plt.title('Total Run by IPL Teams')
plt.xlabel('Teams')
plt.ylabel('Runs')
plt.grid()
plt.show()
                            
```

## TOP 10 RUNS GETER IN IPL

```python
top_batsman=deliveries.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(12,6))
top_batsman.plot(kind='barh',color='orange')
plt.title('Top 10 IPL Batsman')
plt.xlabel('Runs')
plt.ylabel('Players')
plt.grid()
plt.show()
           
```

## MOST SIX HITTER BATSMAN IS IPL

```python
top_six=deliveries[deliveries['batsman_runs']==6]['batsman'].value_counts().head(10)
plt.figure(figsize=(12,6))
top_six.plot(kind='barh',color='purple')
plt.title('Top 10 Most six Hitter batsman in IPL')
plt.xlabel('six')
plt.ylabel('Players')
plt.grid()
plt.show()
    
```

## MOST FOURS HITTER BATSMAN IS IPL

```python
top_fours=deliveries[deliveries['batsman_runs']==4]['batsman'].value_counts().head(10)
plt.figure(figsize=(12,6))
top_fours.plot(kind='barh',color='green')
plt.title('Top 10 Most Fours Hitter batsman in IPL')
plt.xlabel('Fours')
plt.ylabel('Players')
plt.grid()
plt.show()
```

## Top 10 Most Wicket Takers in IPL

```python
top_bowlers=deliveries['bowler'][deliveries['player_dismissed'].notnull()].value_counts().head(10)
plt.figure(figsize=(12,6))
top_bowlers.plot(kind='bar',color='purple')
plt.title('Top 10 Most Wicket Takers in IPL')
plt.xlabel('bowlers')
plt.ylabel('Wickets')
plt.grid()
plt.show()
    
```
